# Data Structures - List, Dictionary, Set & Conditional Statements


List (Creation, Modification and Access):

1. List Creation:

```python
age_list = [24,25,26,27,28]
name_list = ['Shar','Imm','Niranj','Kee','Asi']
print(age_list)
print(name_list)
```

2. List Operations / Modifications:

```python
# a. Append the string "Yazhini" to name_list.
name_list.append('Yazhini')
print('Appended List:',name_list)

# b. Insert the element 30 at index 2 in age_list.
age_list.insert(2,30)
print('Insert age 30 at index 2:',age_list)

# c. Remove the string "Yazhini" from name_list.
name_list.remove('Yazhini')
print('Removed name "Yazhini" from List: ',name_list)

# d. Pop the last element from age_list.
age_list.pop(-1)
print('Delete last element:', age_list)

# e. Extend the age_list with additional ages [29, 30, 26].
age_list.extend([29,30,26])
print('Extended List:', age_list)

# f. Sort age_list in descending order.
age_list.sort(reverse=True)
print('Sorting list in descending order:', age_list)

# g. Find Max age, Min age and sum of all ages from age_list.
print('Maximum age:', max(age_list) )
print('Minimum age:', min(age_list))
print('Sum age:', sum(age_list))
```

3. Accessing List Elements:

```python
# a. Print the first element of name_list.
print('First element: ', name_list[0])

# b. Print the last element of name_list.
print('Last element: ', name_list[-1])

# c. Print the elements from index 2 to index 4 in name_list.
print('Elements from index 2 to 4: ', name_list[2:5])

# d. Print the elements of name_list in reverse order.
print('Name_list in reverse order: ', name_list[::-1])
```

Dictionary (Creation, Modification and Access):

```python
#Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).
student_marks = {'Shar': 91,
                 'Imm': 97,
                 'Niranj' : 85,
                 'Kee' : 65,
                 'Asi' : 46
                 }
print(student_marks)

#Access and print the mark of a specific student, of your choice.
print('Mark of Asina:',student_marks['Asi'])

#Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
student_marks["Janani"] = 80
print('Added new student name:',student_marks)

#Update the mark of any one older student to 82.
student_marks["Kee"] = 82
print('Updated mark of Kee:',student_marks)

#Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.
print('Keys: ', student_marks.keys())
print('values: ', student_marks.values())
print('Items: ', student_marks.items())
```

Sets (Operations):

```python
#Create a set called my_set with following values: ['a','e','i','o','u','a','a','i']
my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)

#Attempt to change the value of my_set[4] = 's'.
my_set[4] = 's'
print(my_set)
   #'set' object does not support item assignment .sets do not have fixed index positions, so we cannot use square brackets to change or assign values.

#Create two sets: set1 with values: {1, 3, 5, 7, 9} set2 with values: {2, 3, 5, 8, 10}
set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}

# Compute and print the union and intersection of set1 and set2
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

Operators & Conditional Statements :

```python
# Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
# Find the performance category based on the input score using following criteria:
# a. Above Average: Score greater than 7
# b. Average: Score between 4 and 7(both inclusive)
# c. Below Average: Score lesser than 4
# Output: Print the Performance category
# Additional Step: You can give a prompt of your choice to each category. For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”.
score = int(input('Enter your score from 0 to 10:' ))
if score > 7:
  print('Above Average:Keep it up')
elif score >=4 and score <=7 :
  print('Average:Capable of good work.')
else:
  print('Below Average:Encouraged to listen carefully.')
```
